# Bitcoin Market Sentiment vs Trader Performance

## Objective

Analyze how trader performance varies across different market sentiment conditions (Fear, Greed, Extreme Fear, Extreme Greed).

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Load Datasets

```python
fg = pd.read_csv("fear_greed_index.csv")
trades = pd.read_csv("historical_data.csv")
```

## Dataset Overview


```python
print("Fear & Greed Shape:", fg.shape)
print("Trades Shape:", trades.shape)

fg.head()
```

```python
trades.head()
```

```python
fg.info()
trades.info()
```

## Data Cleaning

```python
fg['date'] = pd.to_datetime(fg['date'])
```

```python
trades['Timestamp IST'] = pd.to_datetime(
    trades['Timestamp IST'],
    dayfirst=True
)
```

```python
trades['date'] = trades['Timestamp IST'].dt.date
```

```python
fg['date'] = fg['date'].dt.date
```

## Merge Datasets

```python
df = pd.merge(
    trades,
    fg[['date','classification']],
    on='date',
    how='left'
)
```

```python
df.head()
```

## Missing Value Analysis

```python
df.isnull().sum()
```

### Missing Values

Only 6 records had missing sentiment classifications after merging. These rows were removed because they represented a negligible portion of the dataset.



```python
df[df['classification'].isna()][['date']]
```

```python
df = df.dropna(subset=['classification'])
```

```python
df.isnull().sum()
```

## Sentiment Distribution

```python
df['classification'].value_counts()
```

## Profitability Analysis

```python
df.groupby(
    'classification'
)['Closed PnL'].mean()
```

## Win Rate Analysis

```python
#Creating Win/Loss Column
df['Win'] = df['Closed PnL'] > 0
```

```python
df.groupby(
    'classification'
)['Win'].mean() * 100
```

## Trade Size Analysis

```python
df.groupby(
    'classification'
)['Size USD'].mean()
```

## Top Trader Analysis

```python
top_traders = (
    df.groupby('Account')
      ['Closed PnL']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)
```

```python
top_traders.plot(kind='bar')

plt.title("Top 10 Traders by Total PnL")
plt.ylabel("Total Profit")
plt.show()
```

## PnL Distribution

```python
df.groupby(
    'classification'
)['Closed PnL'].mean().plot(
    kind='bar'
)

plt.title("Average Profit by Sentiment")
plt.ylabel("Average PnL")
plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.boxplot(
    data=df,
    x='classification',
    y='Closed PnL'
)

plt.title("PnL Distribution by Sentiment")
plt.show()
```

# Key Findings

1. Extreme Greed periods generated the highest average trader profits.

2. Win rates were highest during Extreme Greed conditions.

3. Extreme Fear periods showed weaker trader performance.

4. Trading activity was concentrated in Fear and Greed market conditions.

# Recommendations

- Monitor market sentiment before entering trades.
- Reduce risk exposure during Extreme Fear periods.
- Consider sentiment-aware trading strategies.
